```python
import pandas as pd
import numpy as np
import ast

from sklearn.neighbors import NearestNeighbors
from sklearn.metrics.pairwise import cosine_similarity, cosine_distances


import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

from collections import defaultdict
from surprise import Dataset, Reader, KNNWithMeans, accuracy
from surprise.model_selection import GridSearchCV
from surprise.model_selection import train_test_split
from surprise.model_selection import KFold

from surprise import SVDpp

import ast
```

```python
data = []
def read_file(filename, data):
    with open(filename, 'r') as file:
        for line in file:
            dict_data = ast.literal_eval(line)
            data.append(dict_data)
```

```python
read_file('/Users/davemag9/VSCode/Machine Learning/lab4/australian_users_items (1).json', data)
df = pd.DataFrame(data)
df.head()
```

```python
user_id ,itemsCount, item_id, item_name, raw_ratings, playtime_2weeks = [],[],[],[],[],[]
for user in data:
    uid = user['user_id']
    iCount = user['items_count']
    for item in user['items']:
        itemsCount.append(iCount)
        user_id.append(uid)
        item_id.append(item['item_id'])
        item_name.append(item['item_name'])
        raw_ratings.append(item['playtime_forever'])
        playtime_2weeks.append(item['playtime_2weeks'])

review = list(zip(user_id ,itemsCount, item_id, item_name, raw_ratings, playtime_2weeks))
```

```python
df = pd.DataFrame({'user':list(d[0] for d in review), 'itemsCount':list(d[1] for d in review),
                           'item_id':list(d[2] for d in review),'item_name':list(d[3] for d in review),
                            'raw_ratings':list(d[4] for d in review), 'playtime_2weeks':list(d[5] for d in review)})

df.head(100)
```

```python
reader = Reader(rating_scale=(0, 1000))
data = Dataset.load_from_df(df[['item_id', 'item_name','raw_ratings']], reader)

trainset, testset = train_test_split(data, test_size=0.3)
```

```python
algo = SVDpp(cache_ratings=False)
algo.fit(trainset)

predictions = algo.test(testset)
rmse = accuracy.rmse(predictions)
mae = accuracy.mae(predictions)
```
